```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
npy_root_path = '../../experiments/data/npy/labn'
```

```python
labn_pred = np.load(f'{npy_root_path}/labn_pred.npy')
labn_true = np.load(f'{npy_root_path}/labn_target.npy')

ofa_pred = np.load(f'{npy_root_path}/OneForAll_pred.npy')
ofa_true = np.load(f'{npy_root_path}/OneForAll_target.npy')
```

```python
labn_pred[:,0]
```

```python
date_range = pd.date_range(start='2023-01-01', end='2023-12-31', periods=len(ofa_true))

for i in range(5):
    plt.figure(figsize=(10, 6))
    plt.plot(date_range, ofa_pred[:,i], label='Predictions', color='red', linestyle='-')
    plt.plot(date_range, ofa_true[:,i], label='Ground Truth', color='blue')
    plt.xlabel('time', fontsize=14)
    plt.ylabel('wave height (m)', fontsize=14)
    plt.title('Time only MLP', fontsize=14)
    # plt.legend(fontsize=18)

    plt.ylim([-5, 5])

    plt.show()
```
